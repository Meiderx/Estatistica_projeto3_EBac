Faça a leitura dos dados do arquivo CSV:


Altere o código abaixo de acordo com seu diretório.

```python
df = pd.read_csv("MODULO7_PROJETOFINAL_BASE_SUPERMERCADO.csv", delimiter=',')

df.head(10)
```

Os campos do nosso dataframe são:

**Title:** Nome do produto.


**Marca:** A marca do produto.


**Preco_Normal:** O preço em que o produto costuma ser vendido quando não há desconto.


**Preco_Desconto:** O preço vendido após o desconto ser aplicado.


**Preco_Anterior:** Preço em que era comercializado o produto antes do desconto aplicado.


**Desconto:** Total de desconto aplicado.






As colunas que aparecem com valores 0 são para os produtos onde não tivemos descontos aplicados.


As categorias estão em espanhol!

# 1 - Traga a média e a mediana dos preços - coluna Preco_Normal - por categoria de produto.
# Identifique as categorias que parecem ter um valor de média abaixo ou acima da mediana.

```python
media = df['Preco_Normal'].groupby(df['Categoria']).mean().round(2)
mediana = df['Preco_Normal'].groupby(df['Categoria']).median().round(2)

nova_df = pd.DataFrame({'Media': media, 'Mediana': mediana})
nova_df['diferenca'] = nova_df['Media'] - nova_df['Mediana']
nova_df
```

Todas as categorias ficaram abaixo da mediana exceto comidas-preparadas.

Podemos notar também que as categorias lacteos, congelados e frutas tiveram uma mediana bem abaixo da media, o que significa que podemos ter outliers puxando esses valores para cima.

# 2 - Traga o desvio padrão por categoria de produto.
# Qual o comportamento da média e mediana nas categorias com maior desvio?

```python
desvio_padrao= df['Preco_Normal'].groupby(df['Categoria']).std().round(2)
nova_df['Desvio Padrao'] = desvio_padrao
nova_df
```

Desvio padrão das categorias mencionadas acima estão muito mariores que as outras, com destaque para o lacteo que foi a categoria que mais deve diferença entre media e mediana e que tem o maior número de desvio padrão. Algo curioso é que a categoria frutas, que antes foi mencionada como uma das que mais tem diferença, teve um dos menores desvio padrão entre todas as categorias.

# 3 - Plot um boxplot da distribuição do Preco_Normal para a categoria que você identificou que tem o maior desvio padrão. Como é a distribuição desses dados segundo o boxplot? Você identifica muitos outliers?

Dica: Para trazer apenas os dados da categoria que você deseja você pode usar o df.loc[df['Categoria'] == 'CATEGORIA ESCOLHIDA'

```python
lacteos = df.loc[df['Categoria'] == 'lacteos']

fig = px.box(lacteos, y='Preco_Normal', width=600, height=550)
fig.show()
```

A coluna apresenta vários outliers que estão deturpando a analise. Também notamos que a mediana é assimitrica e positiva o que quer dizer que está mais perto do 1° quartil do que do 3°. Para finalizar vemos que o 3° quartil e a máxima estão ocupando a maior parte do dataset.

# 4 - Plote um gráfico de barras onde temos a média de descontos por categoria.

```python
df_desconto = df['Desconto'].groupby(df['Categoria']).mean().round(2).sort_values(ascending=False)
label = df_desconto.index

fig = px.bar(df_desconto, labels=label)
fig.update_layout(
    yaxis_title='Desconto',
    title='Desconto por Categoria',
    width=1000,
    height=400
)
fig.show()

```

# 5 - Plote um gráfico de mapa interativo agrupando os dados por categoria, marca e trazendo a média de desconto.

```python

df_grouped = df.groupby(['Categoria', 'Marca'])['Desconto'].mean().reset_index()
df_grouped = df_grouped[df_grouped['Desconto'] > 0].reset_index().drop('index', axis=1)
df_pivot = df_grouped.pivot(index='Categoria', columns='Marca', values='Desconto')

fig = px.imshow(
    df_pivot,
    labels=dict(x="Marca", y="Categoria", color="Desconto"),
    text_auto='.1f')
fig.show()
```
